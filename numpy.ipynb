mini numerical computing engine using NumPy

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate a 256 x 256 grayscale image
np.random.seed(42)

image = np.random.randint(
    0, 256,
    size=(256, 256),
    dtype=np.uint8
)

plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.colorbar()
plt.show()

print("Shape:", image.shape)
print("Data type:", image.dtype)
print("Memory:", image.nbytes, "bytes")
```

```python
image_float = image.astype(np.float32)

normalized = (
    image_float - image_float.min()
) / (
    image_float.max() - image_float.min()
)

print("Minimum:", normalized.min())
print("Maximum:", normalized.max())
print("Mean:", normalized.mean())
print("Standard Deviation:", normalized.std())
```
