mini numerical computing engine using NumPy

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate a 256 x 256 grayscale image
np.random.seed(42)

image = np.random.randint(
    0, 256,
    size=(256, 256),
    dtype=np.uint8
)

plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.colorbar()
plt.show()

print("Shape:", image.shape)
print("Data type:", image.dtype)
print("Memory:", image.nbytes, "bytes")
```

```python
image_float = image.astype(np.float32)

normalized = (
    image_float - image_float.min()
) / (
    image_float.max() - image_float.min()
)

print("Minimum:", normalized.min())
print("Maximum:", normalized.max())
print("Mean:", normalized.mean())
print("Standard Deviation:", normalized.std())
```

```python
kernel = np.array([
    [ 0, -1,  0],
    [-1,  5, -1],
    [ 0, -1,  0]
], dtype=np.float32)

def convolution(image, kernel):

    h, w = image.shape
    kh, kw = kernel.shape

    output = np.zeros((h-kh+1, w-kw+1))

    for i in range(h-kh+1):
        for j in range(w-kw+1):

            region = image[i:i+kh, j:j+kw]

            output[i, j] = np.sum(region * kernel)

    return output

filtered = convolution(normalized, kernel)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(normalized, cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(filtered, cmap='gray')
plt.title("Sharpened")

plt.show()
```
